# 基准：SWE-bench，GAIA，AgentBench

2026年有三个agent评判基准。SWE-bench 测试代码补丁。GAIA 测试通才型工具使用。AgentBench 测试多环境推理。知道他们的组成、他们的污染情况、以及他们没测什么。

## 问题描述

排行榜告诉你哪个模型在某个基准上赢了。但是没有告诉你的事
- 基准是否被污染了（解决方案在训练数据里，测试泄漏）
- 基准是否测量了你关心的事情（代码 vs 浏览 vs 通才）
- 评估器是否稳健（AST匹配、状态检查、人工审查）

在你引用任何数字之前，知道这三个锚定基准以及它们的失败模式。

|基准|构成/原理|污染情况|
|---|---|---|
|SWE-bench|Github仓库，issues的代码修复|issues大多在模型的训练数据库中|
|GAIA|推理、多模态、网页、工具使用等，用来衡量通才能力|...|
|AgentBench|代码、网页、游戏、开放式生成，多轮|...|

### 这些基准没有评估的东西

- 真实世界操作的开销（tokens，时间成本）
- 对抗条件下的安全行为
- 在你的领域下的表现
- 尾部失败（基准返回平均，但是产品关心最差的那1%）

### 什么时候基准失效
- 单一数字执念。  SWE-bench 50% 告诉你的比 P50/P75/P95成本和步数分布少。
- 被污染的声称。 报SWE-bench 但是不提Verified 和 SWE-bench+是误导性的。
- 基准作为开发目标。 针对基准做优化会偏移生产有用性。
