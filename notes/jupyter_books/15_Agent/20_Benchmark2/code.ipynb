# 基准： WebArena 和 OSWorld

WebArea 测试网页agent 在四个自托管应用上的能力，OSWorld测试桌面agent在Ubuntu、Windows和MacOS上的能力。发布的时候都展示出了人与agent的巨大差异。现在差距在缩小，但是失败模式没有发生变化。

## 失败模式

1. GUI锚定。 像素到元素的映射。模型在1920x1080分辨率下很难定位UI元素的位置。
2. 操作性知识。 哪个菜单里面有哪个设置、哪个快捷键、哪个偏好面板。人类用多年攒出来的知识长尾。

## 基准失败模式

- 只看截图的评估。  OSWorld是基于屏幕截图的，但是agent可以是DOM或者API的。
- 忽略轨迹长度。  只看最后成功成功与否，但是消耗了比人工多2倍的操作。
- 过时的自托管应用。  WebArena 的四个应用限制。
