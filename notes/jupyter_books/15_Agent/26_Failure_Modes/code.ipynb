# 失败模式————为什么agents会崩

MASFT 将14种多agents失败归类为三种。Microsoft的分类文档记录了已存在的AI失败是怎么被agents放大的。行业内把agents失败归因总结为五种：幻觉行为、范围蔓延、级联错误、上下文丢失、工具错误。

## 问题描述

在90%场景下能成功的agent，剩余的10%不是随机噪声，而是固定的几种失败模式。一旦你可以叫出它们的名字，你就能够做监视并修复。

## 基本概念

### MASFT（Multi-Agent System Failure Taxonomy）

核心声明：这些失败是多agent系统的基础设计缺陷，而不是LLM带来的限制。因此没办法通过更好大模型修复。

### 五种行业分类

- 幻觉行为。  Agent 尝试调用根本不存在的工具或者使用编造的参数。
- 范围蔓延。  Agent 拓展到用户任务以外。
- 级联错误。  上游错误导致下游错误。
- 上下文丢失。 长上下文，通用问题
- 工具误用。  错误的参数，甚至错误的工具

级联是最麻烦的。

### 缓解措施————每步后面加门禁

在每个推理链后加上自动的验证门禁，检查事实是否锚定在当前环境状态中。

### 什么时候监视失败

- 只看崩溃。  大多数agent失败产生看得过去的输出，你需要上下文层级的检查。
- 没有基线。  漂移检测需要上一个已知的好答案，没有它你不能说新的答案比较差。
- 过度警告。  每个失败都生成一页报告，对其聚类然后限制。
