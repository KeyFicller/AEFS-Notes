## What is Machine Learning

### Core concept

Instead of writing rules, you give the computer thousands of labeled data and let it figure out the rules on it's own.

Shift from **Programming Rules** to **Learing from data**

### Type of Machine Learning

* Supervised Learning: You have input-output pairs.
* Un-supervised Learning: You have inputs only, No labels.
* Reinforcement Learning: Take actions and recieves rewards or penalities.

### Missons of Machine Learning

* Classifcation.
* Regression

### Datasets

* Training Data: Model learns from this data
* Validation Data: Tune hyperparameters, compare models
* Test: Final unbiased performance estimate

### Fitting

* Underfitting: The model is too simple to capture the patterns in the data. eg. **Training error is high, test error is high**

* Overfitting: The model is too complex and memorizes the training data, including it's noise. eg. **Training error is low, test error is high**

* Good fit: The model captures real patterns without memorizing noies. eg. **Training error and test error are bot reasonably low**
