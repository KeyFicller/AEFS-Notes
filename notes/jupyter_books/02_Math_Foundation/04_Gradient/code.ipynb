## Derivative

### Write a numerical derivative calculation

```python
def numeric_derivative(f, x, h = 1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

def f(x):
    return x ** 2

for x in [-2, -1, 0, 1, 2]:
    print(f"Numerical: {numeric_derivative(f, x)}, Analytical: {2 * x}")
```

### Partial derivative and Gradient

```python
def numerical_gradient(f, point, h=1e-5):
    gradient = []
    for i in range(len(point)):
        point_plus = list(point)
        point_minus = list(point)
        point_plus[i] += h
        point_minus[i] -= h
        grad = (f(point_plus) - f(point_minus)) / (2 * h)
        gradient.append(grad)
    return gradient
    
def f_multi(point):
    x, y = point
    return x ** 2 + 3 * x * y + y ** 2

gradient = numerical_gradient(f_multi, [1.0, 2.0])

print("Numercial Gradient at [1, 2]: ", gradient)
print("Analytical Gradient at [1, 2]: ", [2 * 1 + 3 * 2, 3 * 1 + 2 * 2])
```

### Gradient desend with fixed learning ratio

```python
x = 5.0
lr = 0.1

for step in range(20):
    grad = 2 * x
    x = x - lr * grad
    print(f"Step {step+1}: x = {x:.4f}, grad = {grad:.4f}")
```

```python
def f_2d(point):
    x, y = point
    return x ** 2 + y ** 2

point = [3.0, 4.0]
lr = 0.1

for step in range(30):
    grad = numerical_gradient(f_2d, point)
    point = [p - lr * g for p, g in zip(point, grad)]
    loss = f_2d(point)

    if step % 5 == 0 or step == 29:
        print(f"Step {step+1}: point = {point}, loss = {loss:.6f}")
```
