# Matrix Transform

## Rotation

```python
import torch
import math

theta = 45
rotation_2d = torch.tensor([
    [math.cos(theta), -math.sin(theta)], [math.sin(theta), math.cos(theta)]
])

rotation_3d_by_x = torch.tensor([
    [1, 0, 0],
    [0, math.cos(theta), -math.sin(theta)],
    [0, math.sin(theta), math.cos(theta)]
])

rotation_3d_by_y = torch.tensor([
    [math.cos(theta), 0, -math.sin(theta)],
    [0, 1, 0],
    [-math.sin(theta), 0, math.cos(theta)]
])

rotation_3d_by_z = torch.tensor([
    [math.cos(theta), -math.sin(theta), 0],
    [math.sin(theta), math.cos(theta), 0],
    [0, 0, 1]
])
```

## Scale

```python
import torch

# This means scale X component by 0.6, and scale Y component by 2.0
scale_2d = torch.tensor([
    [0.6, 0.0], [0.0, 2.0]
])
```

## Shear

```python
import torch

"""
|                       /
|       Shear into     /
|                     /
"""
shear_2d = torch.tensor([
    [1, 2], [0, 1]
])

print(f"Shear (1, 0) is: {shear_2d @ torch.tensor([1, 0])}")
print(f"Shear (0, 1) is: {shear_2d @ torch.tensor([0, 1])}")
```

## Reflection

```python
import torch

"""
      |
*     |      *
      |
  Reflect Axis
"""
reflect_2d = torch.tensor([
    [-1, 0], [0, 1]
])

print(f"Reflect (1, 0) is: {reflect_2d @ torch.tensor([1, 0])}")
```

## Order does matters

Result differs from:
1. first rotation, then scale
2. first scale, then rotation

```python
base = torch.tensor([3.0, 4.0])

print(f"Order1 is: {
    rotation_2d.to(torch.float) @ 
    scale_2d.to(torch.float) @ 
    shear_2d.to(torch.float) @ 
    reflect_2d.to(torch.float) @
    base}")

print(f"Order2 is: {
    scale_2d.to(torch.float) @ 
    rotation_2d.to(torch.float) @ 
    shear_2d.to(torch.float) @ 
    reflect_2d.to(torch.float) @ 
    base}")
```

## Eigen Vector and Eigen Value

绝大多数向量被矩阵作用后会改变方向。特征向量很特别：矩阵只缩放它们，从不旋转它们。那个缩放因子就是特征值。

```python
A = torch.tensor([
    [2, 1], [1, 2]
]).to(torch.float)

eigen_values, eigen_vectors = torch.linalg.eig(A)

eigen_values = eigen_values.to(torch.float)
eigen_vectors = eigen_vectors.to(torch.float)

print(f"Eigen values: {eigen_values}")
print(f"Eigen vectors: {eigen_vectors}")

print(f"eigen_vectors[:, 0] = {eigen_vectors[:, 0]}")
print(f"A @ eigen_vectors[:, 0] = {A @ eigen_vectors[:, 0]}")
print(f"Scaled factor = {eigen_values[0]}")
```

## Eigen Decomposition

```python
"""

A = V @ D @ V^(-1)

V = matrix whose columns are eigenvectors
D = diagonal matrix of eigenvalues
V^(-1) = inverse of V

"""
```

特征值为什么重要

PCA。 协方差矩阵的特征向量就是主成分。特征值告诉你每个成分捕获了多少方差。按特征值排序，保留前 k 个，你就得到了降维。

稳定性。 在循环网络和动力系统里，模大于 1 的特征值会让输出爆炸。模小于 1 则让它们消失。这就是用一句话说清楚的梯度消失/爆炸问题。

谱方法。 图神经网络用邻接矩阵的特征值。谱聚类用拉普拉斯矩阵的特征值。特征向量揭示了图的结构。

行列式作为体积缩放因子

变换矩阵的行列式告诉你它把面积（二维）或体积（三维）缩放了多少。

# Code Practice

### Custom Transform Matrix

```python
import math

def rotation_2d(theta):
    c, s = math.cos(theta), math.sin(theta)
    return [[c, -s], [s, c]]

def scaling_2d(x, y):
    return [[x, 0], [0, y]]

def shearing_2d(x, y):
    return [[1, x], [1, y]]

def reflection_x():
    return [[1, 0], [0, -1]]

def reflection_y():
    return [[-1, 0], [0, 1]]

def mat_vec_mul(matrix, vector):
    return [
        sum(matrix[i][j] * vector[j] for j in range(len(vector)))
        for i in range(len(matrix))
    ]

def mat_mul(a, b):
    rows_a, cols_b = len(a), len(b[0])
    cols_a = len(a[0])

    return [
        [
            sum(a[i][k] * b[k][j] for k in range(cols_a))
            for j in range(cols_b)
        ]
        for i in range(rows_a)
    ]

point = [1.0, 0.0]
angle = math.pi / 4

rotated = mat_vec_mul(rotation_2d(angle), point)
print(f"Rotate (1,0) by 45 deg: ({rotated[0]:.4f}, {rotated[1]:.4f})")

scaled = mat_vec_mul(scaling_2d(2, 3), [1.0, 1.0])
print(f"Scale (1,1) by (2,3): ({scaled[0]:.1f}, {scaled[1]:.1f})")

sheared = mat_vec_mul(shearing_2d(1, 0), [1.0, 1.0])
print(f"Shear (1,1) kx=1: ({sheared[0]:.1f}, {sheared[1]:.1f})")

reflected = mat_vec_mul(reflection_y(), [2.0, 1.0])
print(f"Reflect (2,1) across y: ({reflected[0]:.1f}, {reflected[1]:.1f})")
```

### Combination of Matrix

```python
R = rotation_2d(math.pi / 2)
S = scaling_2d(2, 0.5)

# Caution with orders
rotation_then_scale = mat_mul(S, R)
scale_then_rotation = mat_mul(R, S)

point = [1.0, 0.0]
result1 = mat_vec_mul(rotation_then_scale, point)
result2 = mat_vec_mul(scale_then_rotation, point)

print(f"Rotation 90 then sacle: ({result1[0]:.2f}, {result1[1]:.2f})")
print(f"Scale then Rotation 90: ({result2[0]:.2f}, {result2[1]:.2f})")

print(f"Same? {result1 == result2}")
```

### Calculation of Determinant

```python
from sympy import discriminant


def eigen_values_2x2(matrix):
    a, b = matrix[0]
    c, d = matrix[1]

    trace = a + d
    det = a * d - b * c
    discriminant = trace ** 2 - 4 * det

    if discriminant < 0:
        real = trace / 2
        imag = (-discriminant) ** 0.5 / 2
        return (complex(real, imag), complex(real, -imag))
    sqrt_disc = discriminant ** 0.5
    return ((trace + sqrt_disc) / 2, (trace - sqrt_disc) / 2)

def eigen_vectors_2x2(matrix, eigen_value):
    a, b = matrix[0]
    c, d = matrix[1]

    if abs(b) > 1e-10:
        v = [b, eigen_value - a]
    elif abs(c) > 1e-10:
        v = [eigen_value - d, c]
    else:
        if abs(a - eigen_value) < 1e-10:
            v = [1, 0]
        else:
            v = [0, 1]
    mag = (v[0] ** 2 + v[1] ** 2) ** 0.5
    return [v[0]/mag, v[1]/mag]

A = [[2, 1], [1, 2]]

vals = eigen_values_2x2(A)
print(f"Matrix: {A}")
print(f"Eigenvalues: {vals[0]:.4f}, {vals[1]:.4f}")

for val in vals:
    vec = eigen_vectors_2x2(A, val)
    result = mat_vec_mul(A, vec)
    scaled = [val * vec[0], val * vec[1]]
    print(f"  lambda={val:.1f}, v={[round(x,4) for x in vec]}")
    print(f"    A@v = {[round(x,4) for x in result]}")
    print(f"    l*v = {[round(x,4) for x in scaled]}")
```

### Interactive Matrix Transform Lab

拖动滑块实时预览变换（网格 / 基向量 / 单位正方形 / 特征向量）。  
组合顺序：**Scale → Shear → Rotate**。  
图内标签用英文（避免 matplotlib 中文乱码）；中文说明在下方信息框。  
依赖：`torch` / `matplotlib` / `ipywidgets`。


```python
%matplotlib inline
import math
import torch
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import ipywidgets as widgets


def make_grid(n=9, span=1.2):
    xs = torch.linspace(-span, span, n)
    ys = torch.linspace(-span, span, n)
    lines = []
    for x in xs:
        lines.append(torch.stack([torch.full_like(ys, x), ys]))
    for y in ys:
        lines.append(torch.stack([xs, torch.full_like(xs, y)]))
    return lines


def compose_matrix(theta_deg, sx, sy, kx, ky):
    """A = R @ Shear @ Scale"""
    th = math.radians(theta_deg)
    c, s = math.cos(th), math.sin(th)
    R = torch.tensor([[c, -s], [s, c]], dtype=torch.float32)
    Shear = torch.tensor([[1.0, kx], [ky, 1.0]], dtype=torch.float32)
    Scale = torch.tensor([[sx, 0.0], [0.0, sy]], dtype=torch.float32)
    return R @ Shear @ Scale


def fmt_c(z, digits=3):
    re, im = float(z.real), float(z.imag)
    if abs(im) < 1e-8:
        return f"{re:.{digits}f}"
    sign = "+" if im >= 0 else "-"
    return f"{re:.{digits}f}{sign}{abs(im):.{digits}f}i"


def eigen_info(matrix):
    evals, evecs = torch.linalg.eig(matrix.to(torch.float32))
    all_real = bool(torch.all(torch.abs(evals.imag) < 1e-6).item())
    return evals, evecs, all_real


def draw_transform(ax, matrix, span=2.5):
    matrix = matrix.to(torch.float32)
    for line in make_grid():
        pts = matrix @ line
        ax.plot(pts[0].numpy(), pts[1].numpy(), color="0.75", linewidth=0.8)

    square = torch.tensor([[0, 1, 1, 0, 0], [0, 0, 1, 1, 0]], dtype=torch.float32)
    sq = matrix @ square
    ax.plot(sq[0].numpy(), sq[1].numpy(), color="tab:blue", linewidth=2.2, label="unit square")

    e1 = (matrix @ torch.tensor([1.0, 0.0])).numpy()
    e2 = (matrix @ torch.tensor([0.0, 1.0])).numpy()
    ax.arrow(0, 0, float(e1[0]), float(e1[1]),
             head_width=0.08, head_length=0.12, length_includes_head=True,
             color="tab:red", linewidth=2, label="e1 -> Ae1")
    ax.arrow(0, 0, float(e2[0]), float(e2[1]),
             head_width=0.08, head_length=0.12, length_includes_head=True,
             color="tab:green", linewidth=2, label="e2 -> Ae2")

    # ========== draw eigenvectors ==========
    evals, evecs, all_real = eigen_info(matrix)
    if all_real:
        colors = ("#7b2cbf", "#e8590c")
        for i in range(2):
            lam = float(evals[i].real)
            v = evecs[:, i].real
            n = float(torch.linalg.norm(v))
            if n < 1e-8:
                continue
            v = (v / n).numpy()
            ax.plot([-span * v[0], span * v[0]], [-span * v[1], span * v[1]],
                    color=colors[i], linewidth=1.8, alpha=0.55, linestyle="--",
                    label=f"eigenvector v{i+1}")
            tip = lam * v
            ax.annotate("", xy=(float(tip[0]), float(tip[1])), xytext=(0, 0),
                        arrowprops=dict(arrowstyle="->", color=colors[i], lw=2.5))
            ax.text(float(tip[0]) * 1.08, float(tip[1]) * 1.08,
                    f"lam{i+1}={lam:.2f}", color=colors[i], fontsize=10, fontweight="bold")
    # ========== end eigenvectors ==========

    ax.axhline(0, color="0.4", linewidth=0.6)
    ax.axvline(0, color="0.4", linewidth=0.6)
    ax.set_xlim(-span, span)
    ax.set_ylim(-span, span)
    ax.set_aspect("equal", adjustable="box")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper right", fontsize=8)
    return evals, evecs, all_real


preset = widgets.Dropdown(
    options=[
        ("Identity", "identity"),
        ("Rotation 45", "rot45"),
        ("Scale (0.6, 2.0)", "scale"),
        ("Shear kx=1", "shear"),
        ("Reflect y-axis", "reflect"),
        ("Custom", "custom"),
    ],
    value="scale",
    description="Preset",
)
theta = widgets.FloatSlider(value=0, min=-180, max=180, step=1, description="theta(deg)", continuous_update=True)
sx = widgets.FloatSlider(value=0.6, min=-2.0, max=2.0, step=0.05, description="sx", continuous_update=True)
sy = widgets.FloatSlider(value=2.0, min=-2.0, max=2.0, step=0.05, description="sy", continuous_update=True)
kx = widgets.FloatSlider(value=0.0, min=-2.0, max=2.0, step=0.05, description="kx", continuous_update=True)
ky = widgets.FloatSlider(value=0.0, min=-2.0, max=2.0, step=0.05, description="ky", continuous_update=True)

info = widgets.HTML()
plot_out = widgets.Output()

PRESETS = {
    "identity": (0, 1.0, 1.0, 0.0, 0.0),
    "rot45": (45, 1.0, 1.0, 0.0, 0.0),
    "scale": (0, 0.6, 2.0, 0.0, 0.0),
    "shear": (0, 1.0, 1.0, 1.0, 0.0),
    "reflect": (0, -1.0, 1.0, 0.0, 0.0),
}


def apply_preset(*_):
    if preset.value == "custom" or preset.value not in PRESETS:
        return
    t, a, b, c, d = PRESETS[preset.value]
    # 批量改滑块时暂停通知，避免一次 Preset 触发 5 次 redraw
    with theta.hold_trait_notifications(), \
         sx.hold_trait_notifications(), \
         sy.hold_trait_notifications(), \
         kx.hold_trait_notifications(), \
         ky.hold_trait_notifications():
        theta.value, sx.value, sy.value, kx.value, ky.value = t, a, b, c, d
    redraw()


def redraw(*_):
    A = compose_matrix(theta.value, sx.value, sy.value, kx.value, ky.value)
    title = (
        f"A = R({theta.value:.0f}) * Shear(kx={kx.value:.2f}, ky={ky.value:.2f}) "
        f"* Scale({sx.value:.2f}, {sy.value:.2f})"
    )

    det = float(torch.linalg.det(A))
    evals, evecs, all_real = eigen_info(A)
    a = A.tolist()
    rows = []
    for i in range(2):
        if all_real:
            vr = evecs[:, i].real
            n = float(torch.linalg.norm(vr))
            if n > 1e-8:
                vr = vr / n
            vtxt = f"[{float(vr[0]):+.3f}, {float(vr[1]):+.3f}]"
        else:
            vtxt = f"[{fmt_c(evecs[0, i])}, {fmt_c(evecs[1, i])}]"
        rows.append(
            f"<tr><td style='padding:2px 10px'><b>λ{i+1}</b> = {fmt_c(evals[i])}</td>"
            f"<td style='padding:2px 10px'><b>v{i+1}</b> = {vtxt}</td></tr>"
        )

    if abs(det) < 1e-6:
        note = "<span style='color:#c00'>奇异：面积塌缩</span>"
    elif not all_real:
        note = "<span style='color:#a60'>复特征值：图上不画实特征方向（例如纯旋转）</span>"
    else:
        note = "<span style='color:#2b8a3e'>实特征值：紫/橙虚线为特征向量方向</span>"

    info.value = (
        "<div style='font-family:Menlo,monospace;font-size:13px;line-height:1.5;"
        "border:1px solid #888;padding:8px;margin:4px 0'>"
        f"<div><b>A</b> = [[{a[0][0]:+.3f}, {a[0][1]:+.3f}], "
        f"[{a[1][0]:+.3f}, {a[1][1]:+.3f}]] &nbsp; <b>det</b> = {det:+.3f}</div>"
        f"<table>{''.join(rows)}</table>"
        f"<div>{note}</div></div>"
    )

    # 建图 / show / close 都在 Output 内，避免重复刷图
    with plot_out:
        clear_output(wait=True)
        fig, ax = plt.subplots(figsize=(6, 6))
        draw_transform(ax, A)
        ax.set_title(title, fontsize=11)
        plt.show()
        plt.close(fig)


sliders = (theta, sx, sy, kx, ky)
preset.observe(apply_preset, names="value")
for w in sliders:
    w.observe(redraw, names="value")

ui = widgets.VBox([
    widgets.HTML("<b>Interactive Matrix Transform Lab</b> — 拖动滑块即时预览（plt.show）"),
    preset, *sliders,
    info,
    plot_out,
])
display(ui)
redraw()  # 初始只画 1 次（不要再调 apply_preset，避免重复）
```
