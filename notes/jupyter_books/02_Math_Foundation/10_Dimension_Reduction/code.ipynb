## Dimension reduction


**The curse of dimensionality**

* Distance becomes meaningless
* Volume concentrates in corners
* Exponentially more data

### PCA: find the directions that matters

**finds the axes along which your data varies the most**
**Rotates your coordinate system so the first axis captures the most variance, the second captures the next most, and so on**

```mermaid
flowchart
A[Center the data -- subtract the mean from each feature] --> B[Compute covariance -- how features move together] --> C[Eigen decomposition -- find the principal directions] --> D[Sort by eigen values -- biggest variance first] --> E[Project -- keep top k eigen vectors, drop the rest]
```

**Why eigen decomposition ?**

The covariance matrix is symmetric and positive semi-definite. It's eigen vectors are orthogonal directions in the feature space. The eigen values tell you how much variance each direction captures.
**The eigen vector with the largest eigen value points aloing the direction of maximum variance**

#### Explained variance ratio

**Each principal component captures a fraction of the total variance. The explained variance ratio tells you how much**

````python
import pandas as pd

df = pd.DataFrame({
    "Component": ["PC1", "PC2", "PC3", "PC4"],
    "Eigen Value": [4.73, 2.51, 1.12, 0.89],
    "Explained Ratio": [0.473, 0.251, 0.112, 0.089],
    "Cumulative Ratio": [0.473, 0.724, 0.836, 0.925]
})

df
````

#### Ways of choosen number of components

* **Threshold.** Keep enough components to explain 90%~95% of the variance.
* **Elbow method.** Plot explianed variance per component. Look for a sharp drop-off
* **Downstream performance.** Use PCA as preprocessing. Sweep k and measure your model's accuaracy. The best k is wherever accuracy plateaus.

### T-SNE: Perserve neighborhoods
**Designed for visualiation. It maps high-dimensional data to 2D(or 3D) while preserving which points are near each other**

In the original space, compute a probability distribution over pairs of points base on their distance. Near points get high probability, far point get low probability. Then find a 2D arrangement where the same probability distribution holds. Points that were neighbors in 784 dimensions stay neighbors in 2D.

**Key properties**
* Non-linear.
* Stochastic
* Perplexity parameter controls how many neighbors to consider
* Distances between clusters in the output are not meaningful.
* Slow on large datasets.
