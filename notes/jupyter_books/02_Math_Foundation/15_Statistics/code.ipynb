## Statistics in ML

### Decriptive Statistices

#### Measures of central tendency

* Mean
`mu = sum(x_i) * (1 / n)`

* Median:
`x[len(x) // 2] or (x[len(x) // 2] + x[len(x) // 2 + 1]) * 0.5`

* Mode
**most frequent value**


#### Measures of spread

* Variance
`sigma^2 = sum((x_i - mu)^2) * (1 / n)`

* Standard deviation
`sigma = sqrt(sigma^2)`

* Range
`max - min`

* IQR
**The range of the middle 50% of the data** (eg. Q3-Q1, Q1 = 25%, Q3 = 75%)

### Bessel's correction
**When computing variance from a sample, divide by (n - 1) instead of n**

### Correlation
**How variables move together, the strength and direction of a linear relationship between two variables**

#### Pearson

`r = sum((x_i - x_mean)(y_i - y_mean)) / (n * sigma_x * sigma_y)`

So. n is to eliminate sqrt(1/n) from both sigma_x and sigma_y

* r = +1: perfect positive linear relationship
* r = -1: perfect negative linear relationship
* r = 0: no linear relationship

Range: [-1, 1]

#### Sperman

1. Replace each value with it's rank (1, 2, 3, ...)
2. Compute the Pearson correlation on the ranks

#### Covariance Matrix

`Cov(X, Y) = (1 / n) * sum((x_i - x_mean)(y_i - y_mean))`

* Cov(X, Y) > 0: X and Y tend to increase together
* Cov(X, Y) < 0: when X increases, Y tends to decrease
* Cov(X, Y) == 0: no linear co-movement
