## Convex Optimization

### Convex function

for any two points x, y in it's domain and any t in [0, 1]:

`f(tx + (1 - t)y) <= tf(x) + (1-t)f(y)`

### Test for convexity

* **Second derivative test (1D)**
    if `f''(x) >= 0` for all x

* **Hessain test (Multivariate)**
    if Hessian Matrix is positive semidefinite for all x

* **Definition test**
    Check for `f(tx + (1 - t)y) <= tf(x) + (1-t)f(y)`

### Core Attribute

**For a convex function, every local minimum is a global minimum.**

### Newton Method

Newton interation:
`x_new = x - H^(-1) * gradient`

Gradient descent iteration:
`x_new = x - lr * gradient`

**Use Hessain matrix instead of learing ratio**

### Constraint Optmization

Minimize f(x) with g(x) = 0

`L(x, lambda) = f(x) + lamba * g(x)`
