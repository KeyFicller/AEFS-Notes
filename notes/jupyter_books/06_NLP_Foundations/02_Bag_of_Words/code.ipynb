# Bag of Words.

Count first, think later.

## Problem Definition

The model needs numbers, you have strings.

How do we turn a variable-length stream of tokens into a fixed-size vector that a classifier can consume.

The dumbest one that works: **Count the words, make a vector.**

## Basic Concept

### Bag of words

Throw away order, vector length is the vocabulary size, position i is the count of word i.

### TF-IDF

A word that appears in every document is uninformative, so scale it down.

A word rare across the corpus but frequent in single document is signal, so scale it up.

```
TF-IDF(w, d) = TF(w, d) * IDF(w) = count(w in d) / |d| * log(N / df(w))

TF: term frequencey in the document
df: document frequency, how many docs contain the word.
N: total documents
log: keeps the weight bounded for ubiquitous words. 
```

# Build your Own

## Build the vocabulary

````python
def build_vocab(docs):
    vocab = {}
    for doc in docs:
        for token in doc:
            if token not in vocab:
                vocab[token] = len(vocab)
    return vocab
````

## Bag of Words

````python
def bag_of_words(docs, vocab):
    matrix = [[0] * len(vocab) for _ in docs]
    for i, doc in enumerate(docs):
        for token in doc:
            matrix[i][vocab[token]] += 1
    return matrix

docs = [["hello", "world"], ["hello", "cat"]]
vocab = build_vocab(docs)
bag_of_words(docs, vocab)
    
````

## Term frequency and document frequency

````python
import math

def term_frequency(doc_bow, doc_length):
    return [c / doc_length if doc_length > 0 else 0 for c in doc_bow]

def document_frequency(bow_matrix):
    df = [0] * len(bow_matrix[0])
    for row in bow_matrix:
        for j, count in enumerate(row):
            if count > 0:
                df[j] += 1
    return df

def inverse_document_frequency(df, n_docs):
    return [math.log((n_docs + 1) / (d + 1)) + 1 for d in df]
````

## TF-IDF

````python
def tfidf(bow_matrix):
    n_docs = len(bow_matrix)
    df = document_frequency(bow_matrix)
    idf = inverse_document_frequency(df, n_docs)
    out = []
    for row in bow_matrix:
        length = sum(row)
        tf = term_frequency(row, length)
        out.append([tf_j * idf_j for tf_j, idf_j in zip(tf, idf)])
    return out

docs = [["hello", "world"], ["hello", "cat"]]
vocab = build_vocab(docs)
bow_matrix = bag_of_words(docs, vocab)
tfidf(bow_matrix)
````
