# Coreference Resolution

"She called him. He did not answer. The doctor was at lunch". Three references to two people and nobody is named. Coreference resolution figures out who is who.

## Problem definition

Coreference resolution link every expression that refers to the same real-world entity into one cluster. It is the glue between surface level NLP(NER, parsing) and downstream semantics (IE, QA, summarization, KG)

## Basic Concept

Coreference clustering: mentions --> entities

### Mention types

1. Named entity.  "Tim Cook"
2. Nominal.       "the CEO", "the Company"
3. Pronoinal.     "he", "she", "they", "it"
4. Appositive.    "Tim Cook, Apple's CEO"

### Architectures

1. Rule-based

Syntactic-tree-based pronoun resolution using grammer rules. 

2. Mention-pair classifier

For every pair of mentions (m_i, m_j), predict whether they corefer. Cluster by transitive closure.

3. Mention-ranking.

For each mention, rank candidate antecedents. Pick the top.

4. Span-based end-to-end.

Transfromer encoder. Enumerate all candiates spans up to a length cap. Perdict mention scores. Predict antecedent-probability fro each span.

5. Generative

Prompt an LLM: "List every pronoun in the text and its antecedent"
