# Natural Language Inference

## Problem Definition

NLI asks: Given a permise t and a hypothesis h, is h entailed by t, contradicted, or neural(unrelated) ?

## Basic Concept

Three way classification of permise vs hypothesis

"The cat is on the mat."
* **Entailment**.  t-->h  "There is a cat."
* **Contradiction**. t-x->h "There is no cat."
* **Neutral**.       t-?->h "The cat is hungry"



# Build your Own

## Run a pretrained NLI model


```python
from transformers import pipeline

import sys
from pathlib import Path

sys.path.insert(0, str(Path("../../00_Common").resolve()))
from user_tools import SectionPrinter

nli_pipeline = pipeline(
    "text-classification",
    model="cross-encoder/nli-deberta-v3-base",
)

premise = "The cat is sleeping on the couch."
hypothesis = "There is a cat in the room."

result = nli_pipeline({
    "text": premise,
    "text_pair": hypothesis,
})
if isinstance(result, list):
    result = result[0]

with SectionPrinter("Run a pretrained NLI model"):
    print(result)


```

## Zero-shot classification

```python
zs = pipeline(
    "zero-shot-classification",
    model = "cross-encoder/nli-deberta-v3-base",
)

text = "The stock market rallied after central bank cut interest rate."
labels = ["finance", "sports", "politics", "technology"]

result = zs(text, candidate_labels=labels)

with SectionPrinter("Zero-shot classification"):
    print(result)

```
