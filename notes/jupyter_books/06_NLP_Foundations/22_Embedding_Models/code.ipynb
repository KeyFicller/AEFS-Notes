### Basic Concept

#### Dense embeddings

One vector per passage.

#### Sparse embeddings

SPLADE-style. A trainsformer predicts a weight for every vocab token, then zeros out most of them. Result is a sparse vector of size |vocab|. **Strong on keyword-heavy queries**

#### Multi-vector

One vector per token. Scoring with MaxSim: for each query token, find the most similar document token, sum the scores.

#### BGE-M3

All three at once, then scores fuse via weighted sum. (2026 default)

#### Matryoshka Representation Learning

Trained so the first N dimensions of the vector form a useful standalone embedding.
