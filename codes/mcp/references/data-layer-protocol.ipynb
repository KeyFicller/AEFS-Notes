## Discover Request

```python
{
    "jsonrpc": "2.0",
    "id": 1,
    "method": "server/discover",
    "params": {
        # Stateless requires each JSON-RPC request to include a _meta field.
        "_meta": {
            # If version rejected, `UnsupportedProtocolVersionError`
            "io.modelcontextprotocol/protocolVersion": "2026-07-28",
            # Identification and versioning information.
            "io.modelcontextprotocol/clientInfo": {
                "name": "Example-Client",
                "version": "1.0.0"
            },
            # The client declares its capabilities on EVERY request.
            "io.modelcontextprotocol/clientCapabilities": {
                "elicitation": {}
            }
        }
    }
}
```

## Discover Response

```python

{
    "jsonrpc": "2.0",
    "id": 1,
    "result": {
        "resultType": "complete",
        "supportedVersions": ["2026-07-28"],
        "capabilities": {
            # The server supports the tools primitive and can honor
            # a `toolsListChanged` filter in `subscription/listen`.
            "tools": {
                "listChaged": true
            },
            # The server also supports the resources primitive.
            # can handle `resources/list` and `resources/read` methods.
            "resources": {}
        },
        "_meta": {
            "io.modelcontextprotocol/serverInfo": {
                "name": "Example-Server",
                "version": "1.0.0"
            }
        },
        # Freshness hint: this result may be reused for 6 minutes.
        "ttlMs": 360000,
        # Who may reuse the cached value: "public" = shareable across
        # users when authorization permits ("private" binds to the caller).
        "cacheScope": "public"
    }
}
```

## Listen Request

```python
"""
为了让Client 接收到Server 的更新，需要使用 `subscription/listen` 方法。
"""

{
    #...
    "method": "subscription/listen",
    "params": {
        # ...
        "notifications": [
            "toolsListChanged": true            
        ]
    }
    #...
}
```

## Acknowledgement

```python
"""
Client 需要对Server 的更新进行确认，使用 `subscription/acknowledge` 方法。
"""

{
    # ...
    "method": "subscription/acknowledge",
    "params": {
        # ...
        "_meta": {
            # Subscription id Tagging
            "io.modelcontextprotocol/subscriptionId": 4
        },
        "notifications": [
            "toolsListChanged": true
        ]
    }
    # ...
}
```

## Notification

```python
"""
Server 需要向Client 发送更新，使用 `notification/xxx/xxx` 方法。
"""

{
    #...

    ## No need for `id`
    "method": "notifications/tools/listchanged",
    "params": {
        # ...
        "_meta": {
            "io.modelcontextprotocol/subscriptionId": 4
        }
    }
}
```
