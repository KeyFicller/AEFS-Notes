## 2. 一次 backward

已有需要梯度的 `x`。令 `y = x ** 2`，对 `y` 做反向传播。

打印 `y`，再打印 `x.grad`（对 `x` 的梯度）。


```python
import torch

x = torch.tensor(3.0, requires_grad=True)

# 作答

import torch

y = x ** 2
y.backward()

print(y)
print(x.grad)
#评阅
# 对。y=9，x.grad=6（2x）。

#参考答案
# y = x ** 2
# y.backward()
# print(y)
# print(x.grad)
```

## 3. 叶子与非叶子

已有叶子张量 `a`，以及由它算出来的 `b = a * 2`。

依次打印：`a.is_leaf`、`b.is_leaf`、`a.requires_grad`、`b.requires_grad`。


```python
import torch

a = torch.tensor(1.5, requires_grad=True)
b = a * 2

# 作答

import torch

# if a requires_grad is not false, then b is a leaf node also.
print(a.is_leaf)
print(b.is_leaf)
print(a.requires_grad)
print(b.requires_grad)
#评阅
# 打印对：True False True True。注释反了——b 由运算得到，不是叶子；requires_grad 会传下来，和 is_leaf 不是一回事。

#参考答案
# print(a.is_leaf, b.is_leaf, a.requires_grad, b.requires_grad)
```

## 4. 梯度会累加

对同一个 `x` 做两次：`y = x * 2` 再 `y.backward()`。

两次之间**不要**清零梯度。打印最终的 `x.grad`。


```python
import torch

x = torch.tensor(1.0, requires_grad=True)

# 作答

import torch
y = x * 2
y.backward()
print(x.grad)

y = x * 2
y.backward()
print(x.grad)
#评阅
# 对。两次不清零，最终 grad 累加到 4。多打一次中间值无妨。

#参考答案
# y = x * 2
# y.backward()
# y = x * 2
# y.backward()
# print(x.grad)
```
