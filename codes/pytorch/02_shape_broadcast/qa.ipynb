## 2. view 改形状

`t` 是连续的一维张量。用 `view` 把它变成 `(2, 3)`，打印结果及其 `shape`。

## 3. permute 调维序

`x` 的形状是 `(2, 3, 4)`。用 `permute` 把维序改成「原第 1 维、原第 0 维、原第 2 维」，即目标形状 `(3, 2, 4)`。

打印 permute 之后的 `shape`。


```python
import torch

x = torch.arange(24).reshape(2, 3, 4)

# 作答

import torch

perm = x.permute(1, 0, 2)
print(x.shape)
print(perm.shape)
#评阅
# 对。permute(1, 0, 2) 得到 (3, 2, 4)。多打了原 shape 无妨。

#参考答案
# print(x.permute(1, 0, 2).shape)
```

## 4. 广播相加

`a` 形状 `(3, 1)`，`b` 形状 `(3,)`。计算 `a + b`（依赖广播），打印结果，再打印结果的 `shape`。


```python
import torch

a = torch.ones(3, 1)
b = torch.arange(3.0)

# 作答
# 3 x 1 
# 1 x 3 
# align to 3 x 3
sum = a + b
print(sum)
print(sum.shape)
#评阅
# 对。广播到 (3, 3)，结果也对。变量名 sum 会盖掉内置函数，下次换个名字更好。

#参考答案
# out = a + b
# print(out)
# print(out.shape)
```

## 5. 不连续后再 view

`y` 来自转置，一般**不连续**。先打印 `y.is_contiguous()`。

再得到一份与 `y` 同值、但是连续的张量，打印它的 `is_contiguous()`。

最后对这份连续张量用 `view` 变成形状 `(12,)`，打印最终 `shape`。


```python
import torch

x = torch.arange(12).reshape(3, 4)
y = x.t()

# 作答

print(y.is_contiguous())

print(y.contiguous())
print(y.contiguous().is_contiguous())

print(y.contiguous().view(12))
print(y.contiguous().view(12).shape)
#评阅
# 对。先看不连续，再 contiguous，最后 view(12)。多次调用 contiguous() 可以，赋给变量一次更干净。

#参考答案
# print(y.is_contiguous())
# z = y.contiguous()
# print(z.is_contiguous())
# print(z.view(12).shape)
```
