## 1. 计数

对 `words` 计数。打印出现次数最多的前两个 `(词, 次数)`（次数相同按词的出现先后由 Counter 决定即可）。


```python
words = ["a", "b", "a", "c", "b", "a"]

# 作答

from collections import Counter

counter = Counter(words)

for word, count in counter.most_common(2):
    print(word, count)
#评阅
# 对。Counter + most_common(2) 就是取前两名。

#参考答案
# from collections import Counter
# print(Counter(words).most_common(2))
```

## 2. 缺省列表

把 `pairs` 按键归到列表里：同一个键的值依次 append。最后打印键 `fruit` 对应的列表。


```python
pairs = [("fruit", "apple"), ("fruit", "pear"), ("color", "red")]

# 作答

from collections import defaultdict

dd = defaultdict(list)

for key, value in pairs:
    dd[key].append(value)

print(dd["fruit"])
#评阅
# 对。defaultdict(list) 省掉了「键不存在就先建空列表」。

#参考答案
# from collections import defaultdict
# bag = defaultdict(list)
# for key, value in pairs:
#     bag[key].append(value)
# print(bag["fruit"])
```

## 3. 缺省整数累加

用缺省值为 `0` 的映射，把 `keys` 里每个键的出现次数累加。打印键 `x` 的次数。


```python
keys = ["x", "y", "x", "x"]

# 作答

from collections import defaultdict

counter = defaultdict(int)

for key in keys:
    counter[key] += 1

print(counter["x"])
#评阅
# 对。defaultdict(int) 缺省是 0，直接 += 1。

#参考答案
# from collections import defaultdict
# counts = defaultdict(int)
# for key in keys:
#     counts[key] += 1
# print(counts["x"])
```

## 4. 双端队列：左边出

用双端队列装入 `items`（保持顺序）。从左边弹出两个元素，依次打印；再打印队列里剩下的内容（转成 list 打印）。


```python
items = ["a", "b", "c", "d"]

# 作答

from collections import deque

dq = deque(items)

for _ in range(2):
    print(dq.popleft())

print(list(dq))
#评阅
# 对。左边出用 popleft，剩下转 list 打印。

#参考答案
# from collections import deque
# q = deque(items)
# print(q.popleft())
# print(q.popleft())
# print(list(q))
```

## 5. 定长队列

做一个最多只保留 3 个元素的双端队列。按顺序放入 `1, 2, 3, 4`。打印最终队列转成的 list（应只剩最后三个）。


```python
# 作答

from collections import deque

dq = deque(maxlen=3)

dq.extend([1, 2, 3, 4])

print(list(dq))
#评阅
# 对。maxlen=3 时再塞第 4 个，左边自动挤掉。

#参考答案
# from collections import deque
# q = deque(maxlen=3)
# for n in (1, 2, 3, 4):
#     q.append(n)
# print(list(q))
```

## 6. 从文件统计词频

读 `box/note.txt`，按空白拆成词并计数。打印出现次数最多的那个词（只要词本身，不要次数）。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "note.txt"
path.write_text("hi hi bye hi ok bye\n", encoding="utf-8")

# 作答

from collections import Counter

counter = Counter(path.read_text(encoding="utf-8").split())

print(counter.most_common(1)[0][0])
#评阅
# 对。split 后交给 Counter，most_common(1) 取第一名的词。

#参考答案
# from collections import Counter
# counts = Counter(path.read_text(encoding="utf-8").split())
# print(counts.most_common(1)[0][0])
```
