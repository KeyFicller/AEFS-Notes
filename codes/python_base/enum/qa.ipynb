## 1. 声明枚举

声明枚举 `Color`，成员 `RED = 1`、`GREEN = 2`。打印 `Color.RED` 的名字和值。


```python
# 作答
from enum import Enum

class Color(Enum):
    RED = 1
    GREEN = 2
    ANOTHER_RED = 1

print(Color.RED.name, Color.RED.value)
print(Color.RED)
print(Color.RED == 1) # 枚举成员不能直接比较
print(Color.RED == Color.RED)
print(Color.RED == Color.ANOTHER_RED)
#评阅
# 对。name / value 取对了。后面别名和 == 的试验也成立：同值成员相等，成员不等于裸整数。

#参考答案
# from enum import Enum
# class Color(Enum):
#     RED = 1
#     GREEN = 2
# print(Color.RED.name, Color.RED.value)
```

## 2. 遍历成员

`Status` 已经声明好。按定义顺序打印每个成员的名字。


```python
from enum import Enum

class Status(Enum):
    TODO = "todo"
    DONE = "done"

# 作答

for status in Status:
    print(status.name)
#评阅
# 对。直接 for 枚举类即可按定义顺序遍历成员。

#参考答案
# for member in Status:
#     print(member.name)
```

## 3. 用值找回成员

`Level` 已经声明好。用值 `2` 找回对应成员，打印该成员的名字。


```python
from enum import Enum

class Level(Enum):
    LOW = 1
    HIGH = 2

# 作答

print(Level(2).name)
#评阅
# 对。Level(2) 按值找回成员。

#参考答案
# print(Level(2).name)
```

## 4. 自动编号

用自动编号声明枚举 `Role`，成员依次是 `ADMIN`、`USER`。打印 `Role.USER` 的值（应是整数）。


```python
# 作答

from enum import Enum, auto

class Role(Enum):
    ADMIN = auto()
    USER = auto()

print(Role.ADMIN.value)
print(Role.USER.value)
#评阅
# 对。auto() 从 1 起自动编号，USER 的值是 2。

#参考答案
# from enum import Enum, auto
# class Role(Enum):
#     ADMIN = auto()
#     USER = auto()
# print(Role.USER.value)
```

## 5. 比较成员

`Mode` 已经声明好。判断变量 `current` 是否等于 `Mode.ON`，打印 `True` 或 `False`。


```python
from enum import Enum

class Mode(Enum):
    OFF = 0
    ON = 1

current = Mode.ON

# 作答

print(current == Mode.ON)
#评阅
# 对。成员之间用 == 比较。

#参考答案
# print(current == Mode.ON)
```

## 6. 把枚举值写入文件

`Priority` 已经声明好。把 `Priority.HIGH.value` 写成一行文本，存进 `box/priority.txt`。再读回文件全文并打印。


```python
from enum import Enum

class Priority(Enum):
    LOW = "low"
    HIGH = "high"

box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "priority.txt"
path.unlink(missing_ok=True)

# 作答

path.write_text(Priority.HIGH.value, encoding="utf-8")

print(path.read_text(encoding="utf-8"))
#评阅
# 对。写入的是 .value（字符串 high），读回正确。

#参考答案
# path.write_text(Priority.HIGH.value, encoding="utf-8")
# print(path.read_text(encoding="utf-8"))
```
