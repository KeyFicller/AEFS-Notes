## 1. 笛卡尔积

列出 `colors` 与 `sizes` 的所有两两组合（每个组合是长度为 2 的元组）。转成 list 后打印。


```python
colors = ["r", "g"]
sizes = ["S", "M"]

# 作答

from itertools import product

print([(x, y) for x, y in product(colors, sizes)])
#评阅
# 对。product 就是笛卡尔积。直接 list(product(...)) 即可，不必再写一层推导。

#参考答案
# from itertools import product
# print(list(product(colors, sizes)))
```

## 2. 排列

对 `letters` 生成所有长度为 2 的排列（顺序不同算不同）。转成 list 后打印。


```python
letters = ["a", "b", "c"]

# 作答

from itertools import permutations

print([(x, y) for x, y in permutations(letters, 2)])
#评阅
# 对。permutations(..., 2) 生成长度为 2 的排列。

#参考答案
# from itertools import permutations
# print(list(permutations(letters, 2)))
```

## 3. 接成一条

把 `a`、`b`、`c` 三条序列按顺序接成一条迭代。转成 list 后打印。


```python
a = [1, 2]
b = (3,)
c = "xy"

# 作答

from itertools import chain

print(list(chain(a, b, c)))
#评阅
# 对。chain 按顺序把多条迭代接成一条。

#参考答案
# from itertools import chain
# print(list(chain(a, b, c)))
```

## 4. 连续相同分组

`items` 已经按值排好序。按连续相同的值分组：对每一组打印 `(值, 这一组的个数)`。


```python
items = ["a", "a", "b", "b", "b", "a"]

# 作答

from itertools import groupby

for key, group in groupby(items):
    print(key, len(list(group)))
#评阅
# 对。groupby 按「连续相同」分组，所以最后单独的 a 是第三组。注意 group 迭代器要立刻用掉（如 list/len）。

#参考答案
# from itertools import groupby
# for key, group in groupby(items):
#     print(key, len(list(group)))
```

## 5. 只取一段

`nums` 可能很长。只取出从下标 `2` 开始的连续 `3` 个元素（不要先把整个序列转成 list 再切片）。转成 list 后打印。


```python
nums = range(10)

# 作答

from itertools import islice

print(list(islice(nums, 2, 5)))
#评阅
# 对。islice(nums, 2, 5) 从下标 2 取到 5（不含），正好 3 个。

#参考答案
# from itertools import islice
# print(list(islice(nums, 2, 5)))
```

## 6. 拼文件行

读 `box/a.txt` 和 `box/b.txt` 的每一行（去掉行尾换行）。按先 a 后 b 的顺序接成一条，转成 list 后打印。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
(box / "a.txt").write_text("one\ntwo\n", encoding="utf-8")
(box / "b.txt").write_text("three\n", encoding="utf-8")

# 作答

from itertools import chain

print(
    list(chain(
        (box / "a.txt").read_text(encoding="utf-8").splitlines(),
        (box / "b.txt").read_text(encoding="utf-8").splitlines(),
    ))
)
#评阅
# 对。splitlines 去掉换行，再用 chain 先 a 后 b。

#参考答案
# from itertools import chain
# lines = chain(
#     (box / "a.txt").read_text(encoding="utf-8").splitlines(),
#     (box / "b.txt").read_text(encoding="utf-8").splitlines(),
# )
# print(list(lines))
```
