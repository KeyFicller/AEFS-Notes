## 1. 声明并构造

声明一个数据类 `Point`，字段是 `x: int` 和 `y: int`。构造 `Point(3, 4)`，打印它（直接 print 实例即可）。


```python
# 作答

from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p = Point(3, 4)

print(p)
#评阅
# 对。@dataclass 声明字段后，print 会带上类名和字段。

#参考答案
# from dataclasses import dataclass
# @dataclass
# class Point:
#     x: int
#     y: int
# print(Point(3, 4))
```

## 2. 默认值

声明数据类 `User`，字段 `name: str`，以及默认值为 `0` 的 `score: int`。只传名字构造 `User("Ada")`，打印 `score`。


```python
# 作答

from dataclasses import dataclass

@dataclass
class User:
    name : str
    score: int = 0

user = User("Ada")

print(user.score)
#评阅
# 对。有默认值的字段放后面，只传 name 时 score 为 0。

#参考答案
# from dataclasses import dataclass
# @dataclass
# class User:
#     name: str
#     score: int = 0
# print(User("Ada").score)
```

## 3. 可变默认要用 factory

声明数据类 `Bag`，字段 `tags` 是字符串列表，默认应是空列表。构造两个实例 `a`、`b`，只给 `a.tags` 追加 `"x"`。分别打印 `a.tags` 和 `b.tags`（`b` 应仍是空列表）。


```python
# 作答

from dataclasses import dataclass, field

@dataclass
class Bag:
    tags: list[str] = field(default_factory=list)

a = Bag()
a.tags.append("x")
print(a.tags)

b = Bag()
print(b.tags)
#评阅
# 对。可变默认必须用 field(default_factory=list)，两个实例不共享同一个列表。

#参考答案
# from dataclasses import dataclass, field
# @dataclass
# class Bag:
#     tags: list[str] = field(default_factory=list)
# a, b = Bag(), Bag()
# a.tags.append("x")
# print(a.tags)
# print(b.tags)
```

## 4. 转成字典

## 5. 替换字段得到新实例

`p` 已经构造好。得到一个 `x` 换成 `9`、其余不变的新实例，打印新实例；再打印原来的 `p`（应仍是旧值）。


```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p = Point(1, 2)

# 作答

from dataclasses import replace
p2 = replace(p, x = 9)

print(p2)
print(p)
#评阅
# 对。replace 返回新实例，原 p 不变。

#参考答案
# from dataclasses import replace
# print(replace(p, x=9))
# print(p)
```

## 6. 冻结后写入文件

声明一个冻结的数据类 `Config`，字段 `host: str` 和 `port: int`。构造 `Config("localhost", 8080)`，用 `asdict` 转成字典后写成 `box/config.json`（JSON，确保中文可读不是必须）。再读回文件全文并打印。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "config.json"
path.unlink(missing_ok=True)

# 作答

from dataclasses import dataclass, asdict
import json

@dataclass(frozen=True)
class Config:
    host: str
    port: int

config = Config("localhost", 8080)

with open(path, "w", encoding="utf-8") as f:
    json.dump(asdict(config), f)

print(path.read_text(encoding="utf-8"))
#评阅
# 对。frozen=True 冻结；asdict 后 json.dump 写入，读回内容正确。

#参考答案
# from dataclasses import dataclass, asdict
# import json
# @dataclass(frozen=True)
# class Config:
#     host: str
#     port: int
# cfg = Config("localhost", 8080)
# path.write_text(json.dumps(asdict(cfg)), encoding="utf-8")
# print(path.read_text(encoding="utf-8"))
```
