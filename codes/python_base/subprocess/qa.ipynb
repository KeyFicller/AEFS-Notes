# subprocess

对应 `stdlib.md`：启动外部程序，拿到输出和退出码。

笔记本在 `python_base/subprocess/qa.ipynb`。需要读写文件时，工作空间就是这个目录，题目文件落在旁边的子目录里。

先运行下一格，得到 `ROOT`。每题只改 `# 作答` 下面的代码。前置代码不用改。做完自己跑通即可，先不要对答案。


```python
from pathlib import Path

def lab_root() -> Path:
    """qa.ipynb 所在目录，即 python_base/subprocess。"""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if folder.name == "subprocess" and (folder / "qa.ipynb").is_file():
            return folder
        candidate = folder / "codes" / "python_base" / "subprocess"
        if (candidate / "qa.ipynb").is_file():
            return candidate
    return here

ROOT = lab_root()
ROOT
```

## 1. 跑起来，拿到输出和退出码

用参数列表 `argv` 启动这个程序。打印它的标准输出，再打印退出码。


```python
argv = ["python3", "-c", "print('hi')"]

# 作答

import subprocess

ret = subprocess.run(argv, capture_output=True)

print(f"标准输出: {ret.stdout.decode()}")
print(f"退出码: {ret.returncode}")
#评阅
# 对。用参数列表启动，capture_output 拿到标准输出，returncode 是 0。
# decode() 把字节变成字符串。

#参考答案
# import subprocess
# result = subprocess.run(argv, capture_output=True, text=True)
# print(result.stdout)
# print(result.returncode)
```

## 2. 程序失败了

这条命令会以状态码 3 结束。打印它的退出码。这一格要正常结束，不要因为对方失败而中断。


```python
argv = ["python3", "-c", "import sys; sys.exit(3)"]

# 作答

import subprocess

ret = subprocess.run(argv, capture_output=True)

print(f"退出码: {ret.returncode}")
#评阅
# 对。没有 check=True，对方以 3 退出时这一格还在，打印的就是 returncode。

#参考答案
# import subprocess
# result = subprocess.run(argv, capture_output=True, text=True)
# print(result.returncode)
```

## 3. 在指定目录里启动

`box` 里已经有 `note.txt`。在这个目录里启动 `argv`，打印标准输出。


```python
box = ROOT / "q3"
box.mkdir(parents=True, exist_ok=True)
(box / "note.txt").write_text("lab\n")
argv = ["python3", "-c", "print(open('note.txt').read(), end='')"]

# 作答

import subprocess

ret = subprocess.run(argv, capture_output=True, cwd=box)

print(f"标准输出: {ret.stdout.decode()}")
#评阅
# 对。cwd=box，命令在这个目录里打开 note.txt，输出是 lab。

#参考答案
# import subprocess
# result = subprocess.run(argv, capture_output=True, text=True, cwd=box)
# print(result.stdout)
```
