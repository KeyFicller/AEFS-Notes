## 1. 固定一个参数

`greet(greeting, name)` 已经写好。做一个新函数，把 `greeting` 固定成 `"hi"`，只再接收 `name`。用它向 `"Ada"` 打招呼并打印返回值。


```python
def greet(greeting: str, name: str) -> str:
    return f"{greeting}, {name}"

# 作答

from functools import partial

greet_hi = partial(greet, "hi")

print(greet_hi("Ada"))
print(greet_hi(name = "Ada"))
#评阅
# 对。partial(greet, "hi") 按位置固定第一个参数，再传 name 即可。

#参考答案
# from functools import partial
# greet_hi = partial(greet, "hi")
# print(greet_hi("Ada"))
```

## 2. 固定关键字参数

`power(base, exp)` 已经写好。做一个新函数，把 `exp` 固定为 `2`（平方）。计算 `power` 意义下的 `5` 的平方并打印。


```python
def power(base: int, exp: int) -> int:
    return base ** exp

# 作答

from functools import partial

power_of_2 = partial(power, exp=2)

print(power_of_2(5))
print(power_of_2(base=5))
#评阅
# 对。用关键字固定 exp=2，位置参数留给 base，所以 power_of_2(5) 不会撞参。

#参考答案
# from functools import partial
# square = partial(power, exp=2)
# print(square(5))
```

## 3. 缓存重复计算

给 `fib` 加上最多记住 `32` 条结果的缓存。然后打印 `fib(10)`，再打印缓存信息里的命中次数 `hits`（从 `fib.cache_info()` 读取）。


```python
def fib(n: int) -> int:
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

# 作答

from functools import lru_cache

@lru_cache(maxsize = 32)
def fib(n: int) -> int:
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(10))
print(fib.cache_info().hits)
#评阅
# 对。lru_cache(maxsize=32) 包住 fib，再读 cache_info().hits。

#参考答案
# from functools import lru_cache
# @lru_cache(maxsize=32)
# def fib(n: int) -> int:
#     if n < 2:
#         return n
#     return fib(n - 1) + fib(n - 2)
# print(fib(10))
# print(fib.cache_info().hits)
```

## 4. 清空缓存

`slow` 已经带缓存。先调用一次 `slow(1)`，再清空缓存，再打印 `slow.cache_info()` 里的 `currsize`（应是 `0`）。


```python
from functools import lru_cache

@lru_cache(maxsize=16)
def slow(n: int) -> int:
    return n * 2

# 作答

slow(1)
slow.cache_clear()

print(slow.cache_info().currsize)
#评阅
# 对。先调用占缓存，再 cache_clear，currsize 变为 0。

#参考答案
# slow(1)
# slow.cache_clear()
# print(slow.cache_info().currsize)
```

## 5. 归约求和

## 6. 带日志的偏函数

`log_line(level, message)` 会把一行写入 `box/app.log`。做一个只接收 `message` 的函数，把 `level` 固定为 `"INFO"`。用它写入 `"ready"`，然后打印日志文件全文。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "app.log"
path.write_text("", encoding="utf-8")

def log_line(level: str, message: str) -> None:
    with path.open("a", encoding="utf-8") as f:
        f.write(f"{level} {message}\n")

# 作答

from functools import partial

info_line = partial(log_line, "INFO")

info_line("ready")

print(path.read_text(encoding="utf-8"))
#评阅
# 对。partial 固定 level，再写入 ready，日志内容正确。

#参考答案
# from functools import partial
# info = partial(log_line, "INFO")
# info("ready")
# print(path.read_text(encoding="utf-8"))
```
