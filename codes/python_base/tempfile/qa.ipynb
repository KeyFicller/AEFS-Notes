# tempfile

对应 `stdlib.md`：临时文件和目录。

笔记本在 `python_base/tempfile/qa.ipynb`。题目要求放进 `box` 时，把临时文件或目录建在那个目录里。

先运行下一格，得到 `ROOT`。每题只改 `# 作答` 下面的代码。前置代码不用改。做完自己跑通即可，先不要对答案。


```python
from pathlib import Path

def lab_root() -> Path:
    """qa.ipynb 所在目录，即 python_base/tempfile。"""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if folder.name == "tempfile" and (folder / "qa.ipynb").is_file():
            return folder
        candidate = folder / "codes" / "python_base" / "tempfile"
        if (candidate / "qa.ipynb").is_file():
            return candidate
    return here

ROOT = lab_root()
ROOT
```

## 1. 临时文件里写再读

在 `box` 里建一个临时文件，写入 `hi`，再把刚写入的内容读出来并打印。用完即删。


```python
box = ROOT / "q1"
box.mkdir(parents=True, exist_ok=True)

# 作答

import tempfile

with tempfile.NamedTemporaryFile(delete=True, dir=box) as f:
    f.write(b"hi")
    f.seek(0)
    for file in box.iterdir():
        print(file.resolve())
    print(f.read())

for file in box.iterdir():
    print(file.resolve())
#评阅
# 对。写进临时文件再读回，退出 with 后文件被删。文本模式打印出来是 hi，不必用 bytes。

#参考答案
# import tempfile
# with tempfile.NamedTemporaryFile("w+", encoding="utf-8", dir=box, delete=True) as f:
#     f.write("hi")
#     f.seek(0)
#     print(f.read())
```

## 2. 临时目录里放一个文件

在 `box` 里建一个临时目录，在其中写入 `note.txt`，内容是 `ok`。打印 `note.txt` 这个文件名（不要整条路径）。


```python
box = ROOT / "q2"
box.mkdir(parents=True, exist_ok=True)

# 作答

import tempfile

with tempfile.TemporaryDirectory(dir=box) as d:
    with open(Path(d) / "note.txt", "w") as f:
        f.write("ok")
    #print(list(box.walk()))
    for root, _, files in box.walk():
        for file in files:
            print(Path(root) / file)
#评阅
# 文件写进临时目录了，但题目只要文件名 note.txt，不要整条路径。

#参考答案
# import tempfile
# from pathlib import Path
# with tempfile.TemporaryDirectory(dir=box) as d:
#     path = Path(d) / "note.txt"
#     path.write_text("ok", encoding="utf-8")
#     print(path.name)
```

## 3. 指定前缀和后缀

在 `box` 里建一个临时文件，名字以 `lab-` 开头、以 `.txt` 结尾。打印这个文件名（不要整条路径）。


```python
box = ROOT / "q3"
box.mkdir(parents=True, exist_ok=True)

# 作答

import tempfile

with tempfile.NamedTemporaryFile(delete=True, dir=box, prefix="lab-", suffix=".txt") as f:
    print(f.name)
#评阅
# 前缀 lab-、后缀 .txt 对了。f.name 是整条路径，题目只要文件名。用 Path(f.name).name。

#参考答案
# import tempfile
# from pathlib import Path
# with tempfile.NamedTemporaryFile(dir=box, prefix="lab-", suffix=".txt") as f:
#     print(Path(f.name).name)
```

## 6. 两次名字不同

在 `box` 里连续建两个临时文件。打印两个文件名是否不同（应为 `True`）。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)

# 作答

import tempfile

with tempfile.NamedTemporaryFile(delete=True, dir=box) as f1:
    with tempfile.NamedTemporaryFile(delete=True, dir=box) as f2:
        print(f1.name != f2.name)
#评阅
# 对。两次名字不同。

#参考答案
# import tempfile
# with tempfile.NamedTemporaryFile(dir=box) as a, tempfile.NamedTemporaryFile(dir=box) as b:
#     print(a.name != b.name)
```
