```python
from pathlib import Path

def lab_root() -> Path:
    """qa.ipynb 所在目录，即 python_base/sys。"""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if folder.name == "sys" and (folder / "qa.ipynb").is_file():
            return folder
        candidate = folder / "codes" / "python_base" / "sys"
        if (candidate / "qa.ipynb").is_file():
            return candidate
    return here

ROOT = lab_root()
ROOT
```

## 2. 从旁边的目录导入

`box` 里有一个 `hello.py`，其中定义了 `greet()`。这个目录现在不在模块搜索路径里。

让当前进程能导入它，调用 `greet()`，打印返回值。


```python
import sys

box = ROOT / "q2"
box.mkdir(parents=True, exist_ok=True)
(box / "hello.py").write_text('def greet():\n    return "hi"\n')

box_str = str(box)
sys.path[:] = [item for item in sys.path if item != box_str]
sys.modules.pop("hello", None)

# 作答

sys.path.append(str(box.resolve()))
from hello import greet
print(f"调用 greet() 的结果: {greet()}")
#评阅
# 对。append 把目录加到末尾，这次能导入并打出 hi。
# 若环境里已经有同名模块，末尾的目录会被先搜到的那个挡住。要让这个目录优先，用 insert(0, ...)。

#参考答案
# sys.path.insert(0, str(box))
# import hello
# print(hello.greet())
```

## 3. 带状态码结束

以状态码 `2` 结束这一步。接住因此抛出的异常，打印里面的状态码。这样后面的单元格还能继续跑。


```python
import sys

# 作答

try:
    sys.exit(2)
except SystemExit as e:
    print(f"状态码: {e.code}")
#评阅
# 对。sys.exit 抛出 SystemExit，code 就是状态码。接住之后单元格还能继续。

#参考答案
# try:
#     sys.exit(2)
# except SystemExit as exc:
#     print(exc.code)
```
