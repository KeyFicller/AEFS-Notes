对应 `stdlib.md`：环境变量、当前目录、和操作系统打交道。

笔记本在 `python_base/os/qa.ipynb`。需要读写文件时，工作空间就是这个目录，题目文件落在旁边的子目录里。

先运行下一格，得到 `ROOT`。每题只改 `# 作答` 下面的代码。前置代码不用改。做完自己跑通即可，先不要对答案。


```python
from pathlib import Path

def lab_root() -> Path:
    """qa.ipynb 所在目录，即 python_base/os。"""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if folder.name == "os" and (folder / "qa.ipynb").is_file():
            return folder
        candidate = folder / "codes" / "python_base" / "os"
        if (candidate / "qa.ipynb").is_file():
            return candidate
    return here

ROOT = lab_root()
ROOT
```

## 3. 换一下当前目录，再换回来

`box` 已经建好。记下现在的工作目录，切到 `box`，打印切过去之后的工作目录，然后切回原来的目录。


```python
import os

box = ROOT / "q3"
box.mkdir(parents=True, exist_ok=True)

# 作答

os.chdir(box)
print(f"当前工作目录: {os.getcwd()}")
os.chdir(ROOT)
print(f"切回原目录: {os.getcwd()}")
#评阅
# 切到 box 之后的路径是对的。切回去用了 ROOT，不是切之前的工作目录。
# 先把 os.getcwd() 存下来，结束时再 chdir 回去。这次 ROOT 恰好就是当时的目录，所以看起来回来了。

#参考答案
# previous = os.getcwd()
# os.chdir(box)
# print(os.getcwd())
# os.chdir(previous)
```

## 4. 列出这一层的名字

## 5. 一次建好多层

创建 `box/a/b`。这个目录已经在的时候，再创建一次也不要失败。最后打印它在不在。


```python
import os

box = ROOT / "q5"

# 作答

(box / "a" / "b").mkdir(parents=True, exist_ok=True)
(box / "a" / "b").mkdir(parents=True, exist_ok=True) # Duplicate mkdir
print(f"目录是否存在: {(box / 'a' / 'b').exists()}")
#评阅
# 目录建出来了，第二次也没有失败。这里用的是 pathlib 的 mkdir。
# os 里对应的是 makedirs，exist_ok=True 才允许再建一次。

#参考答案
# target = box / "a" / "b"
# os.makedirs(target, exist_ok=True)
# os.makedirs(target, exist_ok=True)
# print(os.path.exists(target))
```

## 6. 把子目录里的文件都找出来

`box` 下面有子目录，也有文件。收集里面所有文件的路径，包含子目录，排序后打印。


```python
import os

box = ROOT / "q6"
(box / "src" / "nested").mkdir(parents=True, exist_ok=True)
(box / "src" / "a.txt").write_text("a\n")
(box / "src" / "nested" / "b.txt").write_text("b\n")
(box / "readme.md").write_text("# hi\n")

# 作答

paths = []
for root, dirs, files in os.walk(box):
    for file in files:
        paths.append(os.path.join(root, file))
print(f"收集所有文件的路径: {sorted(paths)}")
#评阅
# 对。walk 走进子目录，只收了文件，三条路径都在。

#参考答案
# paths = []
# for dirpath, _, filenames in os.walk(box):
#     for name in filenames:
#         paths.append(os.path.join(dirpath, name))
# print(sorted(paths))
```
