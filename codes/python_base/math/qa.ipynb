## 4. 浮点是否接近

`u` 和 `v` 是两次算出的近似值。判断它们是否足够接近（用模块里专门比较接近的函数，不要直接 `==`）。打印 `True` 或 `False`。


```python
u = 0.1 + 0.2
v = 0.3

# 作答

print(math.isclose(u, v))
print(u == v)
#评阅
# 对。isclose 得到 True；顺便对照 == 为 False，正好说明浮点误差。

#参考答案
# import math
# print(math.isclose(u, v))
```

## 5. 角度与弧度

把 `90` 度转成弧度，再对这个弧度求正弦。打印正弦值（应非常接近 `1`）。


```python
# 作答

print(math.radians(90))
print(math.sin(math.radians(90)))
#评阅
# 对。radians(90) 再 sin，得到 1.0。题目只要正弦；多打印弧度不影响。

#参考答案
# import math
# print(math.sin(math.radians(90)))
```

## 6. 把结果写入文件

计算 `sqrt(16)` 与 `factorial(5)`，写成两行文本存进 `box/out.txt`（每行一个数）。再读回文件全文并打印。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "out.txt"
path.unlink(missing_ok=True)

# 作答

path.write_text(
    f"{math.sqrt(16)}\n{math.factorial(5)}",
    encoding="utf-8"
)

print(path.read_text(encoding="utf-8"))
#评阅
# 对。两行分别是 4.0 和 120。

#参考答案
# import math
# path.write_text(f"{math.sqrt(16)}\n{math.factorial(5)}\n", encoding="utf-8")
# print(path.read_text(encoding="utf-8"))
```
