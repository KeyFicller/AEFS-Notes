## 1. 断言相等

`add` 已经写好。写一个继承 `unittest.TestCase` 的测试类，里面一个测试方法，断言 `add(2, 3)` 等于 `5`。运行测试。


````python
def add(a: int, b: int) -> int:
    return a + b

# 作答

import unittest

class TestAdd(unittest.TestCase):
    def test_add(self):
        self.assertEqual(add(2, 3), 5)

# unittest.main(argv=[''], exit=False)
unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestAdd)
)

#评阅
# 对。TestCase + assertEqual，用 TextTestRunner 只跑当前类。

#参考答案
# import unittest
# class TestAdd(unittest.TestCase):
#     def test_add(self):
#         self.assertEqual(add(2, 3), 5)
# unittest.TextTestRunner().run(
#     unittest.defaultTestLoader.loadTestsFromTestCase(TestAdd)
# )
````

## 3. 断言抛出异常

`parse` 已经写好。写测试：断言传入 `"x"` 时抛出 `ValueError`。运行测试。


````python
def parse(text: str) -> int:
    return int(text)

# 作答

import unittest

class TestParse(unittest.TestCase):
    def test_parse(self):
        with self.assertRaises(ValueError):
            parse("x")

# unittest.main(argv=[''], exit=False)
unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestParse)
)

#评阅
# 对。with self.assertRaises(ValueError) 包住会抛异常的调用。

#参考答案
# import unittest
# class TestParse(unittest.TestCase):
#     def test_parse(self):
#         with self.assertRaises(ValueError):
#             parse("x")
# unittest.TextTestRunner().run(
#     unittest.defaultTestLoader.loadTestsFromTestCase(TestParse)
# )
````

## 4. setUp

写一个测试类：在 `setUp` 里把 `self.data` 设为 `[1, 2, 3]`。再写两个测试方法——一个断言长度为 `3`，一个断言总和为 `6`。运行测试。


````python
# 作答

import unittest

class TestData(unittest.TestCase):
    def setUp(self):
        self.data = [1, 2, 3]
        
    def test_length(self):
        self.assertEqual(len(self.data), 3)
        
    def test_sum(self):
        self.assertEqual(sum(self.data), 6)
        
# unittest.main(argv=[''], exit=False)
unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestData)
)

#评阅
# 对。setUp 每个测试方法前都会跑，两个方法各用 self.data。

#参考答案
# import unittest
# class TestData(unittest.TestCase):
#     def setUp(self):
#         self.data = [1, 2, 3]
#     def test_length(self):
#         self.assertEqual(len(self.data), 3)
#     def test_sum(self):
#         self.assertEqual(sum(self.data), 6)
# unittest.TextTestRunner().run(
#     unittest.defaultTestLoader.loadTestsFromTestCase(TestData)
# )
````

## 5. 浮点近似相等

写测试：断言 `0.1 + 0.2` 约等于 `0.3`（不要用精确相等）。运行测试。


````python
# 作答

import unittest

class TestFloat(unittest.TestCase):
    def test_float(self):
        self.assertAlmostEqual(0.1 + 0.2, 0.3)

# unittest.main(argv=[''], exit=False)
unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestFloat)
)

#评阅
# 对。assertAlmostEqual 用于浮点近似比较。

#参考答案
# import unittest
# class TestFloat(unittest.TestCase):
#     def test_float(self):
#         self.assertAlmostEqual(0.1 + 0.2, 0.3)
# unittest.TextTestRunner().run(
#     unittest.defaultTestLoader.loadTestsFromTestCase(TestFloat)
# )
````

## 6. 多组输入

`cases` 是几组 `(a, b, expected)`。写一个测试方法，对每组断言 `add(a, b) == expected`；某一组失败时，输出里要能看出是哪一组。运行测试。


````python
def add(a: int, b: int) -> int:
    return a + b

cases = [(1, 2, 3), (10, -3, 7), (0, 0, 1)]

# 作答

import unittest

class TestAdd(unittest.TestCase):
    def test_add(self):
        for a, b, expected in cases:
            with self.subTest(a=a, b=b):
                self.assertEqual(add(a, b), expected)

# unittest.main(argv=[''], exit=False)
unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestAdd)
)

#评阅
# 对。subTest 让某一组失败时输出里仍能看到 a、b，且其余组还会继续跑。

#参考答案
# import unittest
# class TestAdd(unittest.TestCase):
#     def test_add(self):
#         for a, b, expected in cases:
#             with self.subTest(a=a, b=b):
#                 self.assertEqual(add(a, b), expected)
# unittest.TextTestRunner().run(
#     unittest.defaultTestLoader.loadTestsFromTestCase(TestAdd)
# )

````
