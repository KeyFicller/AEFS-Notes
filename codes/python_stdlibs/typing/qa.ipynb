## 1. 给函数加标注

写一个函数 `double`：参数 `n` 标注为 `int`，返回值标注为 `int`，返回 `n * 2`。调用 `double(21)` 并打印结果。

## 2. 容器元素类型

写一个函数 `join_names`：参数 `names` 标注为「字符串列表」，返回值标注为 `str`，用逗号拼接后返回。调用 `join_names(["Ada", "Bob"])` 并打印。


```python
# 作答

def join_names(names: list[str]) -> str:
    return ",".join(names)

print(join_names(["Ada", "Bob"]))
#评阅
# 对。list[str] 标容器元素类型，join 拼接正确。

#参考答案
# def join_names(names: list[str]) -> str:
#     return ",".join(names)
# print(join_names(["Ada", "Bob"]))
```

## 3. 可以为空

写一个函数 `label`：参数 `text` 可以是 `str` 或 `None`。若是 `None` 返回 `"(empty)"`，否则返回原字符串。分别对 `"hi"` 和 `None` 调用并打印两次结果。


```python
# 作答

def label(text: str | None) -> str:
    return text or "(empty)"

print(label("hi"))
print(label(None))
#评阅
# 对。str | None 表示可空；对本题的 "hi" / None 结果正确。
# 注意：text or "(empty)" 也会把空字符串 "" 当成 empty；若只想拦 None，用 is None 更准。

#参考答案
# def label(text: str | None) -> str:
#     return "(empty)" if text is None else text
# print(label("hi"))
# print(label(None))
```

## 4. TypedDict

用 `TypedDict` 声明 `User`，字段 `name: str` 和 `age: int`。构造一个符合该结构的字典，打印其中的 `name`。


```python
# 作答

from typing import TypedDict

class User(TypedDict):
    name: str
    age: int

user = User(name="Ada", age=21)

print(user["name"])
#评阅
# 对。TypedDict 描述字典形状，用键取值。

#参考答案
# from typing import TypedDict
# class User(TypedDict):
#     name: str
#     age: int
# user: User = {"name": "Ada", "age": 21}
# print(user["name"])
```

## 5. 字面量取值

写一个函数 `traffic`：参数 `color` 只能是字面量 `"red"`、`"yellow"`、`"green"` 之一，返回值是 `str`。传入 `"red"` 时返回 `"stop"`，其它颜色返回 `"go"`。调用 `traffic("red")` 并打印。


```python
# 作答

from typing import Literal

def traffic(
    color: Literal["red", "yellow", "green"]
) -> str:
    return {
        "red": "stop",
        "yellow": "go",
        "green": "go"
    }[color]

print(traffic("red"))
#评阅
# 对。Literal 限定三个颜色，red → stop。

#参考答案
# from typing import Literal
# def traffic(color: Literal["red", "yellow", "green"]) -> str:
#     return "stop" if color == "red" else "go"
# print(traffic("red"))
```

## 6. 读出标注并写入文件

`greet` 已经写好。读出它的类型标注（参数和返回值），写成 JSON 存进 `box/hints.json`（键用字符串形式的类型名即可，例如用 `str(注解)`）。再读回文件全文并打印。


```python
def greet(name: str) -> str:
    return f"hi, {name}"

box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "hints.json"
path.unlink(missing_ok=True)

# 作答

from typing import get_type_hints
import json

hints = get_type_hints(greet)

hints_stringfy = {
    k: str(v) for k, v in hints.items()
}

with path.open("w", encoding="utf-8") as f:
    json.dump(hints_stringfy, f)

print(path.read_text(encoding="utf-8"))
#评阅
# 对。get_type_hints 取出注解，str(...) 后再 json.dump，文件内容正确。

#参考答案
# from typing import get_type_hints
# import json
# hints = {k: str(v) for k, v in get_type_hints(greet).items()}
# path.write_text(json.dumps(hints), encoding="utf-8")
# print(path.read_text(encoding="utf-8"))
```
