## 1. 一个位置参数

参数列表是 `qa.py hello`。解析之后打印名字，结果应是 `hello`。

用给定的列表解析，不要读笔记本真正的启动参数。

`parse_args` 收到的列表里不要带脚本名。`qa.py` 是程序名，不是那个位置参数。


```python
argv = ["qa.py", "hello"]

# 作答

import argparse

parser = argparse.ArgumentParser(prog="qa.py", description="一个位置参数")
parser.add_argument("name", help="位置参数：名字")
args = parser.parse_args(argv[1:])
print(f"你的名字是: {args.name}")
#评阅
# 对。parse_args 用 argv[1:]，脚本名不进参数列表，name 是 hello。

#参考答案
# import argparse
# parser = argparse.ArgumentParser()
# parser.add_argument("name")
# print(parser.parse_args(argv[1:]).name)
```

## 2. 可选项，缺了就用默认值

`--count` 是整数。没写的时候当作 `1`，写了就用写的那个数。

分别解析 `plain` 和 `with_count`，各打印一次 count。


```python
plain = ["qa.py"]
with_count = ["qa.py", "--count", "3"]

# 作答

parser = argparse.ArgumentParser(prog="qa.py", description="一个整数选项")
parser.add_argument("--count", type=int, default = 1)

args_plain = parser.parse_args(plain[1:])
print(f"plain 的 count: {args_plain.count}")

args_with_count = parser.parse_args(with_count[1:])
print(f"with_count 的 count: {args_with_count.count}")
#评阅
# 对。type=int、default=1，没写是 1，写了是 3。
# 这一格没有 import argparse。上一格跑过之后内核里还有这个名字，所以这次能跑；重启后再单独跑这一格会 NameError。

#参考答案
# import argparse
# parser = argparse.ArgumentParser()
# parser.add_argument("--count", type=int, default=1)
# print(parser.parse_args(plain[1:]).count)
# print(parser.parse_args(with_count[1:]).count)
```

## 3. 一个开关

`--verbose` 出现时为真，不出现时为假。

分别解析 `quiet` 和 `verbose`，各打印一次这个开关的值。


```python
quiet = ["qa.py"]
verbose = ["qa.py", "--verbose"]

# 作答

parser = argparse.ArgumentParser(prog="qa.py", description="一个布尔选项")
parser.add_argument("--verbose", action="store_true")

args_quiet = parser.parse_args(quiet[1:])
print(f"quiet 的 verbose: {args_quiet.verbose}")

args_verbose = parser.parse_args(verbose[1:])
print(f"verbose 的 verbose: {args_verbose.verbose}")
#评阅
# 对。action="store_true"：没出现是 False，出现是 True。
# 同样没有 import argparse，单独重跑这一格时要自己导入。

#参考答案
# import argparse
# parser = argparse.ArgumentParser()
# parser.add_argument("--verbose", action="store_true")
# print(parser.parse_args(quiet[1:]).verbose)
# print(parser.parse_args(verbose[1:]).verbose)
```
