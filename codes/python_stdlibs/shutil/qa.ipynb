对应 `stdlib.md`：复制、移动、删除目录树。

笔记本在 `python_base/shutil/qa.ipynb`。需要读写文件时，工作空间就是这个目录，题目文件落在旁边的子目录里。

先运行下一格，得到 `ROOT`。每题只改 `# 作答` 下面的代码。前置代码不用改。做完自己跑通即可，先不要对答案。


```python
from pathlib import Path

def lab_root() -> Path:
    """qa.ipynb 所在目录，即 python_base/shutil。"""
    here = Path.cwd().resolve()
    for folder in [here, *here.parents]:
        if folder.name == "shutil" and (folder / "qa.ipynb").is_file():
            return folder
        candidate = folder / "codes" / "python_base" / "shutil"
        if (candidate / "qa.ipynb").is_file():
            return candidate
    return here

ROOT = lab_root()
ROOT
```

## 1. 复制一个文件

`src.txt` 已经写好。把它复制成同目录下的 `dst.txt`，打印副本里的内容。


```python
box = ROOT / "q1"
box.mkdir(parents=True, exist_ok=True)
src = box / "src.txt"
src.write_text("hi\n")
dst = box / "dst.txt"
dst.unlink(missing_ok=True)

# 作答

import shutil

shutil.copy(src, dst)
print(f"复制后的内容: {dst.read_text()}")
#评阅
# 对。copy 把 src.txt 复制成 dst.txt，内容是 hi。

#参考答案
# import shutil
# shutil.copy(src, dst)
# print(dst.read_text())
```

## 2. 复制整棵目录

`src` 下面有子目录和文件。把它整棵复制到 `dst`。打印 `dst` 里所有文件相对 `dst` 的路径，排序后打印。


```python
box = ROOT / "q2"
src = box / "src"
(src / "nested").mkdir(parents=True, exist_ok=True)
(src / "a.txt").write_text("a\n")
(src / "nested" / "b.txt").write_text("b\n")
dst = box / "dst"
if dst.exists():
    for path in sorted(dst.rglob("*"), reverse=True):
        path.rmdir() if path.is_dir() else path.unlink()
    dst.rmdir()

# 作答

import shutil
import os

shutil.copytree(src, dst)

for root, _, files in os.walk(dst):
    for file in files:
        print(f"相对路径: {(Path(root) / file).relative_to(dst)}")
#评阅
# 复制对了，copytree 把子目录里的文件也带过去了。
# 题目要排序后打印。这里边走边打印，没有 sorted。这次 walk 的顺序刚好是 a.txt、nested/b.txt。

#参考答案
# import shutil
# shutil.copytree(src, dst)
# files = sorted(p.relative_to(dst) for p in dst.rglob("*") if p.is_file())
# for path in files:
#     print(path)
```

## 3. 移动到另一个目录

`note.txt` 在 `box` 里，`done` 目录已经建好。把文件移动到 `done/note.txt`。打印新路径在不在，以及原路径还在不在。


```python
box = ROOT / "q3"
(box / "done").mkdir(parents=True, exist_ok=True)
src = box / "note.txt"
src.write_text("move\n")
dst = box / "done" / "note.txt"
if dst.exists():
    dst.replace(src)

# 作答

import shutil

shutil.move(src, dst)
print(f"新路径存在: {dst.exists()}")
print(f"原路径存在: {src.exists()}")
#评阅
# 对。move 之后新路径在，原路径不在。

#参考答案
# import shutil
# shutil.move(src, dst)
# print(dst.exists(), src.exists())
```

## 4. 删掉整棵目录

`tree` 里面有子目录和文件。把这棵目录整个删掉，打印它还在不在。


```python
box = ROOT / "q4"
tree = box / "tree"
(tree / "nested").mkdir(parents=True, exist_ok=True)
(tree / "a.txt").write_text("a\n")
(tree / "nested" / "b.txt").write_text("b\n")

# 作答

import shutil

shutil.rmtree(tree)
print(f"目录是否存在: {tree.exists()}")
#评阅
# 对。rmtree 连里面的文件和子目录一起删掉，tree 不在了。

#参考答案
# import shutil
# shutil.rmtree(tree)
# print(tree.exists())
```
