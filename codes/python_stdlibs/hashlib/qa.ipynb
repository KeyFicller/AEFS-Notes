## 1. SHA-256 十六进制摘要

对字符串 `"hello"`（UTF-8）做 SHA-256，打印十六进制摘要字符串。


```python
text = "hello"

# 作答

import hashlib

print(hashlib.sha256(text.encode(encoding="utf-8")).hexdigest())
#评阅
# 对。encode 成 bytes 后 sha256().hexdigest()。

#参考答案
# import hashlib
# print(hashlib.sha256(text.encode()).hexdigest())
```

## 2. 原始摘要长度

对 `b"abc"` 做 SHA-256。打印 `digest()` 的字节长度（应是整数 `32`）。


```python
data = b"abc"

# 作答

import hashlib

print(
    len(hashlib.sha256(data).digest())
)
#评阅
# 对。digest() 是 32 字节原始摘要。

#参考答案
# import hashlib
# print(len(hashlib.sha256(data).digest()))
```

## 3. 分段 update

新建一个 SHA-256 对象，先 `update(b"hel")`，再 `update(b"lo")`。打印十六进制摘要，应和直接对 `b"hello"` 做 SHA-256 的结果相同。


```python
# 作答

import hashlib

sha = hashlib.sha256()
sha.update(b"hel")
sha.update(b"lo")
print(sha.hexdigest())

print(sha.hexdigest() == hashlib.sha256(b"hello").hexdigest())
#评阅
# 对。分段 update 与一次喂入等价；顺带比对 True 也好。

#参考答案
# import hashlib
# h = hashlib.sha256()
# h.update(b"hel")
# h.update(b"lo")
# print(h.hexdigest())
```

## 4. 不同算法结果不同

对同一段 `b"hello"` 分别做 MD5 和 SHA-256，打印两个十六进制摘要是否相等（应为 `False`）。


```python
data = b"hello"

# 作答

import hashlib

print(
    hashlib.md5(data).hexdigest() 
    ==
    hashlib.sha256(data).hexdigest()
)
#评阅
# 对。不同算法摘要不同。

#参考答案
# import hashlib
# print(hashlib.md5(data).hexdigest() == hashlib.sha256(data).hexdigest())
```

## 5. 改一个字节摘要就变

## 6. 用名字构造算法

不要写 `hashlib.sha256(...)` 这种属性调用。用算法名字符串 `"sha256"` 构造摘要对象，对 `b"ok"` 求十六进制摘要并打印。


```python
data = b"ok"
name = "sha256"

# 作答

import hashlib

print(
    hashlib.new(name, data).hexdigest()
)
#评阅
# 对。hashlib.new(name, data) 按名字构造。

#参考答案
# import hashlib
# print(hashlib.new(name, data).hexdigest())
```
