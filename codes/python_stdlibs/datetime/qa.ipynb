## 1. 构造日期

构造 2024 年 3 月 15 日这个日期。依次打印年、月、日三个整数。


```python
# 作答

import datetime

date = datetime.date(2024, 3, 15)

print(date.year)
print(date.month)
print(date.day)

```

## 2. 加减天数

`start` 是起点。得到 7 天后的日期，打印它的 `isoformat()`。


```python
from datetime import date

start = date(2024, 3, 15)

# 作答

date = start + datetime.timedelta(days=7)

print(date.isoformat())
```

## 3. 格式化成字符串

把 `when` 格式化成 `2024-03-15 09:30` 这种字符串（年月日用 `-`，时分用 `:`，中间一个空格）。打印结果。


```python
from datetime import datetime

when = datetime(2024, 3, 15, 9, 30)

# 作答

print(when.strftime("%Y-%m-%d %H:%M"))
```

## 4. 从字符串解析

`raw` 是 `年-月-日`。解析成日期，打印它的星期几（周一是 `0`，周日是 `6`）。


```python
raw = "2024-03-15"

# 作答
import datetime

date = datetime.datetime.strptime(raw, "%Y-%m-%d")

print(date.weekday())
```

## 5. 相差几天

计算 `end` 比 `start` 晚多少天。打印这个整数天数。


```python
from datetime import date

start = date(2024, 3, 15)
end = date(2024, 4, 1)

# 作答

import datetime
print((end - start).days)
```

## 6. 写入日志时间戳

把 `when` 写成 ISO 格式字符串（中间是 `T`），追加一行到 `box/events.log`。然后读出文件全文并打印。


```python
from datetime import datetime

box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "events.log"
path.write_text("", encoding="utf-8")
when = datetime(2024, 3, 15, 9, 30, 0)

# 作答

with path.open("a", encoding="utf-8") as f:
    f.write(when.isoformat() + "\n")

print(path.read_text(encoding="utf-8"))
```
