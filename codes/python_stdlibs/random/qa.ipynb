## 1. 固定种子后取整数

把种子设为 `42`，再取一个闭区间 `[1, 6]` 里的随机整数（模拟掷骰）。打印这个整数。

## 3. 抽样多个（不放回）

## 4. 原地打乱

## 6. 加权抽取

种子设为 `5`。从 `["A", "B", "C"]` 里按权重 `[1, 1, 8]` 抽 `1` 个，打印抽到的那个元素。


```python
items = ["A", "B", "C"]
weights = [1, 1, 8]

# 作答

import random

random.seed(5)

print(random.choices(items, weights))

#评阅
# 按旧题干：seed + 两个 randint 写入文件，结果可复现，对。
# 已把第 6 题改成模块内的加权抽样（不再绑文件），请按新题干重做本格。

#参考答案（新题）
# import random
# random.seed(5)
# print(random.choices(["A", "B", "C"], weights=[1, 1, 8], k=1)[0])

```
