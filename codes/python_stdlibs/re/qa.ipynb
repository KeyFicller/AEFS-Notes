## 1. 找第一个匹配

`text` 里找第一个连续数字串。打印这段数字；如果没有，打印 `None`。


```python
text = "order-42-32-ready"

# 作答

import re

if match := re.search(r"[0-9]+", text):
    print(match.group())
else:
    print(None)
#评阅
# 对。search 找第一个，group() 取出匹配；没有匹配时打印 None。笔记本里若仍显示 None，再跑一次本格（旧输出）。

#参考答案
# import re
# m = re.search(r"\d+", text)
# print(m.group() if m else None)
```

## 2. 找出全部

找出 `text` 里所有连续的英文单词（只含字母）。按出现顺序打印这个列表。


```python
text = "hi, AI-42; bye"

# 作答

import re

for word in re.findall(r"[a-zA-Z]+", text):
    print(word)
#评阅
# 词找对了，但题目要「打印这个列表」。用 findall 得到列表后 print 整个列表，不要逐个 print。

#参考答案
# import re
# print(re.findall(r"[A-Za-z]+", text))
```

## 3. 取出分组

`text` 形如 `名字:分数`。用带括号的正则取出名字和分数，打印这两个字符串。


```python
text = "小明:90"

# 作答

import re

if match := re.search(r"(\w+):(\d+)", text):
    #print(match[0])
    print(match[1])
    print(match[2])
else:
    print(None)
#评阅
# 对。括号分组后用 match[1]、match[2]（或 group(1)、group(2)）取出名字和分数。

#参考答案
# import re
# m = re.search(r"(.+):(\d+)", text)
# print(m.group(1))
# print(m.group(2))
```

## 4. 替换

把 `text` 里所有连续数字替换成 `#`。打印替换后的字符串。


```python
text = "a12b345c"

# 作答

import re

print(re.sub(r"[0-9]+", "#", text))
#评阅
# 对。sub 一次换掉所有连续数字。

#参考答案
# import re
# print(re.sub(r"\d+", "#", text))
```

## 5. 忽略大小写

判断 `text` 里是否出现单词 `error`（大小写随意）。打印 `True` 或 `False`。


```python
text = "FATAL Error near line 3"

# 作答

import re

if re.search(r"error", text, re.IGNORECASE):
    print(True)
else:
    print(False)
#评阅
# 对。IGNORECASE 忽略大小写。若要真按「单词」边界，可写成 r"\berror\b"。

#参考答案
# import re
# print(bool(re.search(r"\berror\b", text, flags=re.IGNORECASE)))
```

## 6. 从文件里抽邮箱

读 `box/note.txt`。找出所有简单邮箱（形如 `字数字符@字数字符.字数字符`，本地和域名里允许 `.` `_` `-`）。按出现顺序去重后打印列表。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
path = box / "note.txt"
path.write_text(
    "联系 a@x.com 或 b_c@y-z.org，重复 a@x.com 一次。\n",
    encoding="utf-8",
)

# 作答

import re

emails = []

for line in path.read_text(encoding="utf-8").splitlines():
    for email in re.findall(r"[a-zA-Z0-9_.-]+@[a-zA-Z0-9_.-]+", line):
        if email not in emails:
            emails.append(email)

print(emails)
#评阅
# 对。findall + 按出现顺序去重。域名里最好强制有一个点，避免匹配到 a@x 这种。

#参考答案
# import re
# text = path.read_text(encoding="utf-8")
# found = re.findall(r"[\w.-]+@[\w.-]+\.[\w.-]+", text)
# print(list(dict.fromkeys(found)))
```
