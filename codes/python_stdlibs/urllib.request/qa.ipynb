## 1. 打开并读出文本

`note.txt` 已写好。用它的 `file://` 地址打开，按 UTF-8 读出全文并打印。


```python
box = ROOT / "q1"
box.mkdir(parents=True, exist_ok=True)
note = box / "note.txt"
note.write_text("hello\n", encoding="utf-8")
url = note.as_uri()

# 作答

import urllib.request

text = urllib.request.urlopen(url).read().decode(encoding="utf-8")
print(text)
#评阅
# 对。urlopen 读 bytes，再 decode 成文本。

#参考答案
# from urllib.request import urlopen
# print(urlopen(url).read().decode())
```

## 2. 带请求头打开

用 `Request` 打开 `url`，请求头 `User-Agent` 设为 `lab`。读出正文并打印（仍是 `hello`）。


```python
box = ROOT / "q2"
box.mkdir(parents=True, exist_ok=True)
note = box / "note.txt"
note.write_text("hello\n", encoding="utf-8")
url = note.as_uri()

# 作答

import urllib.request

request = urllib.request.Request(url)
request.add_header("User-Agent", "lab")
text = urllib.request.urlopen(request).read().decode(encoding="utf-8")
print(text)
#评阅
# 对。Request + add_header("User-Agent", "lab")，正文仍是 hello。

#参考答案
# from urllib.request import Request, urlopen
# req = Request(url, headers={"User-Agent": "lab"})
# print(urlopen(req).read().decode())
```

## 3. 拼查询串再打开

`base` 是不带查询串的地址。把 `params` 编成查询串接到 `base` 后面（形如 `...?name=Ada&n=1`），用这个完整地址打开，读出正文并打印。


```python
box = ROOT / "q3"
box.mkdir(parents=True, exist_ok=True)
note = box / "note.txt"
note.write_text("ok\n", encoding="utf-8")
base = note.as_uri()
params = {"name": "Ada", "n": 1}

# 作答

import urllib.request, urllib.parse

url = base + "?" + urllib.parse.urlencode(params)
text = urllib.request.urlopen(url).read().decode(encoding="utf-8")
print(text)
#评阅
# 对。urlencode 拼查询串，再 urlopen。

#参考答案
# from urllib.parse import urlencode
# from urllib.request import urlopen
# full = base + "?" + urlencode(params)
# print(urlopen(full).read().decode())
```

## 4. 读响应头

打开 `url`，打印响应头里的 `Content-type`。


```python
box = ROOT / "q4"
box.mkdir(parents=True, exist_ok=True)
note = box / "note.txt"
note.write_text("hello\n", encoding="utf-8")
url = note.as_uri()

# 作答

import urllib.request

response = urllib.request.urlopen(url)
print(response.headers["Content-type"])
#评阅
# 对。响应头 Content-type 是 text/plain。

#参考答案
# from urllib.request import urlopen
# with urlopen(url) as resp:
#     print(resp.headers["Content-type"])
```

## 5. 打不开时接住异常

`url` 指向一个不存在的本地文件。打开它，不要让这一格中断。打印接到的异常类型名。


```python
box = ROOT / "q5"
box.mkdir(parents=True, exist_ok=True)
url = (box / "missing.txt").as_uri()

# 作答

try:
    urllib.request.urlopen(url)
except Exception as e:
    print(type(e).__name__)
#评阅
# 对。类型名是 URLError。更准是只接 URLError，而不是笼统的 Exception。单独跑本格要自己 import urllib.request。

#参考答案
# from urllib.error import URLError
# from urllib.request import urlopen
# try:
#     urlopen(url)
# except URLError as exc:
#     print(type(exc).__name__)
```

## 6. 最终地址

打开 `url`，打印响应的最终地址（应仍是这个 `file://` 地址）。


```python
box = ROOT / "q6"
box.mkdir(parents=True, exist_ok=True)
note = box / "note.txt"
note.write_text("hello\n", encoding="utf-8")
url = note.as_uri()

# 作答

import urllib.request

response = urllib.request.urlopen(url)
print(response.geturl())
#评阅
# 对。geturl() 就是最终地址。

#参考答案
# from urllib.request import urlopen
# with urlopen(url) as resp:
#     print(resp.geturl())
```
